```python
import numpy as np
import matplotlib.pyplot as plt


def sinuz_vektor_doldur(vektor, genlik, faz, frekans=None, periyot=None, fs=1024.0):
    n = len(vektor)
    if n <= 0 or (n & (n - 1)) != 0:
        raise ValueError("Vektör boyutu 2'nin kuvveti olmalıdır.")
    if frekans is None and periyot is None:
        raise ValueError("Frekans veya periyot değerinden en az biri verilmelidir.")
    if frekans is None:
        frekans = 1.0 / periyot

    zaman = np.arange(n, dtype=np.float64) / fs
    vektor[:] = genlik * np.sin(2.0 * np.pi * frekans * zaman + faz)
    return vektor


N = 512
vektor = np.zeros(N, dtype=np.float64)
sinuz_vektor_doldur(vektor, genlik=5.0, faz=np.pi / 4, frekans=10.0, fs=512.0)

print("Vektör uzunluğu:", len(vektor), "İlk 5 değer:", vektor[:5])

plt.figure(figsize=(10, 4))
plt.plot(vektor, marker=".", markersize=3)
plt.title("1. Sinüzoidal Dalga Vektörü (N = 512)")
plt.xlabel("Örnek İndeksi (n)")
plt.ylabel("Genlik")
plt.grid(True)
plt.show()
```
