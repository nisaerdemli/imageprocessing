```python
import os
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# 1. 1000x1000 renkli (3 kanallı) örnek BMP oluştur ve kaydet
renkli_ornek = np.zeros((1000, 1000, 3), dtype=np.uint8)
renkli_ornek[:, :500] = (255, 180, 100)  # Sol yarı (BGR)
renkli_ornek[:, 500:] = (80, 200, 120)   # Sağ yarı (BGR)

renkli_dosya = "renkli_1000.bmp"
cv2.imwrite(renkli_dosya, renkli_ornek)

# 2. Kaydedilen BMP dosyasını aç ve Colab'da göster
img = cv2.imread(renkli_dosya, cv2.IMREAD_COLOR)
cv2_imshow(img)

# 3. Gri tonlamaya dönüştür ve BMP olarak kaydet
gri = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
gri_dosya = "gri_1000.bmp"
cv2.imwrite(gri_dosya, gri)

# 4. Boyut ve disk alanı kıyaslaması
renkli_bayt = os.path.getsize(renkli_dosya)
gri_bayt = os.path.getsize(gri_dosya)

print(f"Renkli shape: {img.shape} | Disk boyutu: {renkli_bayt} bayt ({renkli_bayt / (1024*1024):.2f} MiB)")
print(f"Gri shape:    {gri.shape} | Disk boyutu: {gri_bayt} bayt ({gri_bayt / (1024*1024):.2f} MiB)")
```
