```python
import cv2
import matplotlib.pyplot as plt

# 1. Renkli görüntüyü oku
image = cv2.imread('resim.jpg')

if image is None:
    print("Hata: 'resim.jpg' dosyası bulunamadı!")
else:
    channel_names = ('Mavi (Blue)', 'Yeşil (Green)', 'Kırmızı (Red)')
    colors = ('blue', 'green', 'red')

    plt.figure(figsize=(10, 5))
    plt.title("Görev 1: OpenCV calcHist ile Kanal Histogramları")
    plt.xlabel("Piksel Yoğunluğu (0 - 255)")
    plt.ylabel("Piksel Sayısı")

    # 2. Her kanal için histogram hesapla ve çiz
    for i, (col, name) in enumerate(zip(colors, channel_names)):
        # cv2.calcHist çıktısını tek boyutlu düz diziye çeviriyoruz (.ravel())
        hist = cv2.calcHist([image], [i], None, [256], [0, 256]).ravel()

        # İlk 3 değeri konsola yazdır
        print(f"--- {name} Kanalı (İlk 3 Değer) ---")
        for val in range(3):
            print(f"Piksel Değeri {val}: {int(hist[val])} adet")
        print()

        # Grafiği çizdir
        plt.plot(hist, color=col, label=name)
        plt.xlim([0, 256])

    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.show()
```
