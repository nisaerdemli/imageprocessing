```python
import time
import cv2
import numpy as np

# 8K (7680x4320) 3-kanallı BMP oluştur ve kaydet
img_8k_olustur = np.random.randint(0, 256, size=(4320, 7680, 3), dtype=np.uint8)
cv2.imwrite("ornek_8k.bmp", img_8k_olustur)

# Sadece açma
img_8k = cv2.imread("ornek_8k.bmp", cv2.IMREAD_COLOR)

# Sadece düzenleme (her pikseli 2'ye bölme) süresi
t_basla = time.perf_counter()
duzenlenmis_8k = img_8k // 2
t_bitis = time.perf_counter()

# Kaydetme (ölçüm dışında)
cv2.imwrite("duzenlenmis_8k.bmp", duzenlenmis_8k)

gecen_sn = t_bitis - t_basla
print(f"Piksel bölme süresi: {gecen_sn * 1000:.3f} ms ({gecen_sn:.6f} sn)")
```
