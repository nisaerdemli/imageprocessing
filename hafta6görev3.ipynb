```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np

# 1. Resmi yükleme ve gri tonlamaya çevirme
img = cv2.imread('indir.jpg')
if img is None:
  print('Resim yüklenemedi. Lütfen dosya adını kontrol edin.')
else:
  gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

  # 2. Median Filtre ile ön gürültü temizliği
  median_gray = cv2.medianBlur(gray, 5)

  # 3. Canny Kenar Algılama (İsteğinize uygun olarak ara adım gösterimi)
  edges = cv2.Canny(median_gray, 50, 150)
  median_edges = cv2.medianBlur(edges, 3)

  # 4. OpenCV Hough Circles (Hough Çember Dönüşümü) Uygulama
  # Not: HoughCircles fonksiyonu dahili gradyan hesaplaması yaptığı için
  # genellikle median filtreden geçirilmiş gri tonlamalı 'median_gray' ile en iyi sonucu verir.
  circles = cv2.HoughCircles(
      median_gray,
      cv2.HOUGH_GRADIENT,
      dp=1.2,  # Çözünürlük oranı
      minDist=30,  # Tespit edilen çember merkezleri arasındaki minimum mesafe
      param1=50,  # Canny kenar algılayıcı için üst eşik değeri
      param2=30,  # Akümülatör eşik değeri (küçüldükçe sahte çemberler artar)
      minRadius=10,  # Minimum çember yarıçapı
      maxRadius=150,  # Maksimum çember yarıçapı
  )

  # Çizim için orijinal resmin bir kopyasını alıyoruz
  img_circles = img.copy()

  # 5. Bulunan Çemberleri Resim Üzerine Çizme
  if circles is not None:
    circles = np.uint16(np.around(circles))
    for i in circles[0, :]:
      center = (i[0], i[1])
      radius = i[2]

      # Çemberin merkezini küçük kırmızı bir nokta ile çizin (BGR: 0, 0, 255)
      cv2.circle(img_circles, center, 2, (0, 0, 255), 3)

      # Çemberin çevresini farklı bir renk (Yeşil) ile çizin (BGR: 0, 255, 0)
      cv2.circle(img_circles, center, radius, (0, 255, 0), 2)

    print(f'Toplam {len(circles[0])} adet çember tespit edildi ve çizildi.')
  else:
    print('Hiç çember bulunamadı. Parametreleri resminize göre ayarlayabilirsiniz.')

  # Sonuçları Colab Üzerinde Görselleştirme
  print('Canny + Median Kenar Görüntüsü:')
  cv2_imshow(median_edges)

  print('Çember Tespiti Sonucu (Yeşil Çemberler, Kırmızı Merkezler):')
  cv2_imshow(img_circles)
```
