```python
import numpy as np
import matplotlib.pyplot as plt


def gurultu_ekle(vektor, gurultu_orani, tohum=42):
    if not (0.0 <= gurultu_orani <= 1.0):
        raise ValueError("Gürültü oranı [0, 1] aralığında olmalıdır.")

    rng = np.random.default_rng(tohum)
    genlik = np.max(np.abs(vektor))
    gurultu_genligi = gurultu_orani * genlik
    gurultulu = vektor.astype(np.float64) + rng.uniform(
        -gurultu_genligi, gurultu_genligi, size=len(vektor)
    )
    return gurultulu


N = 512
t = np.linspace(0, 1, N, endpoint=False)
temiz = 8.0 * np.sin(2 * np.pi * 6 * t)
gurultulu = gurultu_ekle(temiz, gurultu_orani=0.2)

plt.figure(figsize=(10, 4))
plt.plot(gurultulu, label="Gürültülü Sinyal (oran=0.2)", alpha=0.85)
plt.plot(temiz, label="Orijinal Sinyal", linestyle="--")
plt.title("2. Genliğe Orantılı Gürültü Eklenmiş Vektör")
plt.xlabel("Örnek İndeksi")
plt.ylabel("Değer")
plt.legend()
plt.grid(True)
plt.show()
```
