```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter1d, median_filter

rng = np.random.default_rng(10)
t = np.linspace(0, 1, 512, endpoint=False)
temiz = 10.0 * np.sin(2 * np.pi * 5 * t)
gurultulu = temiz + rng.normal(0.0, 2.0, size=len(temiz))

# Sırasıyla önce Gauss, sonra Medyan filtre
gauss_cikis = gaussian_filter1d(gurultulu, sigma=2.5)
medyan_cikis = median_filter(gauss_cikis, size=5)

plt.figure(figsize=(10, 7))

plt.subplot(3, 1, 1)
plt.plot(gurultulu)
plt.title("Gürültülü Veri")

plt.subplot(3, 1, 2)
plt.plot(gauss_cikis)
plt.title("Gauss Filtre Sonrası")

plt.subplot(3, 1, 3)
plt.plot(medyan_cikis)
plt.title("Gauss + Medyan Filtre Sonrası")

plt.tight_layout()
plt.show()
```
