```python
import numpy as np


def zncc_hesapla(img1, img2):
    a = img1.astype(np.float64)
    b = img2.astype(np.float64)
    a_sifir_ort = a - np.mean(a)
    b_sifir_ort = b - np.mean(b)
    payda = np.sqrt(np.sum(a_sifir_ort ** 2) * np.sum(b_sifir_ort ** 2))
    if payda == 0:
        return 0.0
    return float(np.sum(a_sifir_ort * b_sifir_ort) / payda)


y_idx, x_idx = np.ogrid[:128, :128]
orijinal = np.zeros((128, 128), dtype=np.uint8)
orijinal[20:108, 20:108] = 180

rng = np.random.default_rng(7)
gurultulu = np.clip(orijinal.astype(np.float64) + rng.normal(0, 25, size=orijinal.shape), 0, 255).astype(np.uint8)

# Gürültülü görüntünün DFT ve Ters DFT dönüşümü
F_gurultulu = np.fft.fft2(gurultulu.astype(np.float64))
ters_dft_img = np.clip(np.real(np.fft.ifft2(F_gurultulu)), 0, 255).astype(np.uint8)

skor_zncc = zncc_hesapla(orijinal, ters_dft_img)
print(f"9. Orijinal vs Ters DFT Görüntü ZNCC Skoru: {skor_zncc:.6f}")
```
