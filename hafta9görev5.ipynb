```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

bgr = cv2.imread("renkli_1000.bmp", cv2.IMREAD_COLOR)
if bgr is None:
    bgr = np.zeros((1000, 1000, 3), dtype=np.uint8)
    bgr[:, :500] = (255, 180, 100)
    bgr[:, 500:] = (80, 200, 120)

tam_kirmizi = (0, 0, 255)  # BGR'de tam kırmızı

# 1) RGB uzayı
rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
R = rgb[:, :, 0].astype(np.int16)
G = rgb[:, :, 1].astype(np.int16)
B = rgb[:, :, 2].astype(np.int16)
mask_rgb = (
    (R > 95) & (G > 40) & (B > 20) &
    ((np.maximum(np.maximum(R, G), B) - np.minimum(np.minimum(R, G), B)) > 15) &
    (np.abs(R - G) > 15) &
    (R > G) & (R > B)
)
sonuc_rgb = bgr.copy()
sonuc_rgb[~mask_rgb] = tam_kirmizi

# 2) YCbCr uzayı
ycrcb = cv2.cvtColor(bgr, cv2.COLOR_BGR2YCrCb)
_, Cr, Cb = cv2.split(ycrcb)
mask_ycbcr = (Cr >= 133) & (Cr <= 173) & (Cb >= 77) & (Cb <= 127)
sonuc_ycbcr = bgr.copy()
sonuc_ycbcr[~mask_ycbcr] = tam_kirmizi

# 3) HSV uzayı
hsv = cv2.cvtColor(bgr, cv2.COLOR_BGR2HSV)
H, S, V = cv2.split(hsv)
mask_hsv = (H <= 25) & (S >= 40) & (S <= 180) & (V >= 60)
sonuc_hsv = bgr.copy()
sonuc_hsv[~mask_hsv] = tam_kirmizi

print("RGB Maske:")
cv2_imshow(sonuc_rgb)
print("YCbCr Maske:")
cv2_imshow(sonuc_ycbcr)
print("HSV Maske:")
cv2_imshow(sonuc_hsv)
```
