```python
import cv2
import numpy as np

# 1. Resmi yükleme ve gri tonlamaya çevirme
# 'resim.jpg' yerine kendi resim dosyanızın yolunu yazabilirsiniz.
img = cv2.imread('resim.jpg')
if img is None:
    print("Resim yüklenemedi. Lütfen dosya yolunu kontrol edin.")
    exit()

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 2. Canny Kenar Algılama
# Eşik değerlerini (thresholds) resminize göre ayarlayabilirsiniz.
edges = cv2.Canny(gray, 50, 150, apertureSize=3)

# 3. Median Filtre Uygulama
# Kenar görüntüsündeki gürültüleri azaltmak için median blur kullanılır (çekirdek boyutu 3 seçilmiştir).
median_edges = cv2.medianBlur(edges, 3)

# Çizimleri üzerine yapabilmek için orijinal resmin bir kopyasını alalım
img_standard = img.copy()
img_probabilistic = img.copy()

# 4. İlk Test: Standart Hough Dönüşümü (cv2.HoughLines)
# Sonsuz uzunlukta doğrular döndürür.
lines = cv2.HoughLines(median_edges, 1, np.pi / 180, 150)

if lines is not None:
    for line in lines:
        rho, theta = line[0]
        a = np.cos(theta)
        b = np.sin(theta)
        x0 = a * rho
        y0 = b * rho
        # Doğruyu uzatmak için büyük bir katsayı kullanıyoruz
        x1 = int(x0 + 1000 * (-b))
        y1 = int(y0 + 1000 * (a))
        x2 = int(x0 - 1000 * (-b))
        y2 = int(y0 - 1000 * (a))
        # Standart Hough doğruları için KIRMIZI renk kullanıyoruz (BGR: 0, 0, 255)
        cv2.line(img_standard, (x1, y1), (x2, y2), (0, 0, 255), 2)

# 5. İkinci Test: Olasılıksal Hough Dönüşümü (cv2.HoughLinesP)
# Doğru parçalarının başlangıç ve bitiş noktalarını doğrudan döndürür.
linesP = cv2.HoughLinesP(
    median_edges, 1, np.pi / 180, 50, minLineLength=50, maxLineGap=10
)

if linesP is not None:
    for line in linesP:
        x1, y1, x2, y2 = line[0]
        # Olasılıksal Hough doğruları için YEŞİL renk kullanıyoruz (BGR: 0, 255, 0)
        cv2.line(img_probabilistic, (x1, y1), (x2, y2), (0, 255, 0), 2)

# Sonuçları Görselleştirme
cv2.imshow("Orijinal Resim", img)
cv2.imshow("Canny + Median Filtre", median_edges)
cv2.imshow("Standart Hough (Kirmizi)", img_standard)
cv2.imshow("Olasiliksal Hough (Yesil)", img_probabilistic)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
