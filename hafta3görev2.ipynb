```python
import cv2
import numpy as np

# 1. Düşük kontrastlı tek kanallı (grayscale) görüntüyü yükleme
# Not: Kendi dosya yolunuzu belirtiniz.
resim_yolu = "dusuk_kontrast.jpg"
img = cv2.imread(resim_yolu, cv2.IMREAD_GRAYSCALE)

if img is None:
    raise FileNotFoundError(f"Görüntü bulunamadı: {resim_yolu}")

yukseklik, genislik = img.shape
toplam_piksel = yukseklik * genislik

# ----------------------------------------------------
# 2. 256 Değerli Histogramı Manuel Çıkarma
# ----------------------------------------------------
histogram = np.zeros(256, dtype=int)

for y in range(yukseklik):
    for x in range(genislik):
        piksel_degeri = img[y, x]
        histogram[piksel_degeri] += 1

# ----------------------------------------------------
# 3. Kümülatif Dağılım Fonksiyonu (CDF) Hesabı
# ----------------------------------------------------
cdf = np.zeros(256, dtype=int)
kumulatif_toplam = 0

for i in range(256):
    kumulatif_toplam += histogram[i]
    cdf[i] = kumulatif_toplam

# ----------------------------------------------------
# 4. Histogram Eşitleme (Histogram Equalization)
# ----------------------------------------------------
# Sıfırdan büyük ilk (en küçük) CDF değerini bulma
cdf_min = 0
for deger in cdf:
    if deger > 0:
        cdf_min = deger
        break

# CDF değerlerini [0, 255] aralığına eşleyen dönüşüm tablosu (Look-Up Table)
esitleme_tablosu = np.zeros(256, dtype=np.uint8)

payda = toplam_piksel - cdf_min
for i in range(256):
    if cdf[i] < cdf_min or payda == 0:
        esitleme_tablosu[i] = 0
    else:
        # Standart histogram eşitleme formülü
        yeni_deger = round(((cdf[i] - cdf_min) / payda) * 255)
        esitleme_tablosu[i] = np.clip(yeni_deger, 0, 255)

# Yeni piksel değerlerini orijinal görüntüye uygulama
esitlenmis_img = np.zeros_like(img)

for y in range(yukseklik):
    for x in range(genislik):
        esitlenmis_img[y, x] = esitleme_tablosu[img[y, x]]

# ----------------------------------------------------
# 5. Çıktıyı Kaydetme
# ----------------------------------------------------
cikti_yolu = "esitlenmis_cikti.jpg"
cv2.imwrite(cikti_yolu, esitlenmis_img)
print(f"İşlem tamamlandı, çıktı kaydedildi: {cikti_yolu}")
```
