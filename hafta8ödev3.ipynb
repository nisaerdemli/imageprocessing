```python
import os
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# 1000x1000 3 kanallı örnek BMP oluştur ve kaydet
renkli = np.zeros((1000, 1000, 3), dtype=np.uint8)
renkli[:, :500] = (255, 180, 100)  # Sol yarı (BGR)
renkli[:, 500:] = (80, 200, 120)   # Sağ yarı (BGR)
cv2.imwrite("renkli_1000.bmp", renkli)

# Aç ve göster
img = cv2.imread("renkli_1000.bmp", cv2.IMREAD_COLOR)
cv2_imshow(img)

# Gri tonlamaya çevir ve BMP kaydet
gri = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imwrite("gri_1000.bmp", gri)

renkli_boyut = os.path.getsize("renkli_1000.bmp")
gri_boyut = os.path.getsize("gri_1000.bmp")

print(f"Renkli -> shape: {img.shape}, disk: {renkli_boyut} bayt ({renkli_boyut / (1024*1024):.2f} MiB)")
print(f"Gri    -> shape: {gri.shape}, disk: {gri_boyut} bayt ({gri_boyut / (1024*1024):.2f} MiB)")
```
