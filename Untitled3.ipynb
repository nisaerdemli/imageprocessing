```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Kendi Gri Ton Dönüşüm Fonksiyonumuz
def custom_bgr_to_gray(img):
    height, width, _ = img.shape
    gray_img = np.zeros((height, width), dtype=np.uint8)

    # OpenCV BGR formatında tutar:
    # B = kanal 0, G = kanal 1, R = kanal 2
    for r in range(height):
        for c in range(width):
            b = int(img[r, c, 0])
            g = int(img[r, c, 1])
            red = int(img[r, c, 2])

            # OpenCV'nin tam sayı (fixed-point) yuvarlama formülü:
            # (4899*R + 9617*G + 1868*B + 8192) >> 14
            val = (4899 * red + 9617 * g + 1868 * b + 8192) >> 14
            gray_img[r, c] = val

    return gray_img

# 2. Resmi oku
image = cv2.imread('resim.jpg')

if image is None:
    print("Hata: 'resim.jpg' bulunamadı!")
else:
    # A) OpenCV hazır fonksiyonu ile griye çevirme
    gray_opencv = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # B) Kendi yazdığımız fonksiyon ile griye çevirme
    print("Kendi gri ton dönüşüm fonksiyonumuz çalışıyor...")
    gray_custom = custom_bgr_to_gray(image)
    print("Dönüşüm tamamlandı!\n")

    # C) Hata Değeri Hesabı (Mutlak Farkların Toplamı)
    # Taşma (overflow) olmaması için int32 tipine çevirip fark alıyoruz
    diff = np.abs(gray_opencv.astype(np.int32) - gray_custom.astype(np.int32))
    total_error = np.sum(diff)

    print("================ RAPOR ================")
    print(f"Toplam Mutlak Hata (Total Absolute Error) : {total_error}")
    print(f"Ortalama Piksel Başına Hata              : {total_error / (image.shape[0] * image.shape[1]):.6f}")
    if total_error == 0:
        print("Tebrikler: Kendi fonksiyonunuz OpenCV ile %100 birebir aynı çıktıyı üretti!")
    else:
        print("Ufak yuvarlama farkı tespit edildi.")
    print("=======================================\n")

    # D) Görselleri yan yana göster
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title("Orijinal Renkli")
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.imshow(gray_opencv, cmap='gray')
    plt.title("OpenCV cvtColor")
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.imshow(gray_custom, cmap='gray')
    plt.title("Kendi Fonksiyonumuz")
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```
