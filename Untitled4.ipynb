```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np

# 1. Resmi yükleme, gri tonlamaya çevirme, Canny ve Median filtre uygulama
img = cv2.imread('indir.jpg')
if img is None:
  print('Resim yüklenemedi. Lütfen dosya adını kontrol edin.')
else:
  gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

  # Canny Kenar Algılama
  edges = cv2.Canny(gray, 50, 150, apertureSize=3)

  # Median Filtre ile gürültü azaltma
  median_edges = cv2.medianBlur(edges, 3)

  # 2. Hough Dönüşümü Parametreleri ve Sin/Cos Ön Hesaplama (Lookup Table)
  height, width = median_edges.shape
  # Görüntünün köşegen uzunluğu (maksimum olası rho değeri)
  diag_len = int(np.ceil(np.sqrt(height**2 + width**2)))

  # Theta açıları: -90 ile +90 derece arası (radyan cinsinden)
  thetas = np.deg2rad(np.arange(-90, 90, 1))
  num_thetas = len(thetas)

  # Rho aralığı: -diag_len ile +diag_len arası
  rhos = np.arange(-diag_len, diag_len + 1, 1)
  num_rhos = len(rhos)

  # İSTEK: Sin ve Cos değerlerini baştan hesaplayıp bellekte tutma (Lookup Table)
  cos_thetas = np.cos(thetas)
  sin_thetas = np.sin(thetas)

  # 3. Oylama Matrisini (Accumulator) Oluşturma
  # Satırlar rho (mesafe), sütunlar theta (açı) değerlerini temsil eder.
  accumulator = np.zeros((num_rhos, num_thetas), dtype=np.uint64)

  # Kenar piksel koordinatlarını alma (y: satır, x: sütun)
  edge_y, edge_x = np.nonzero(median_edges)

  # 4. Oylama İşleminin Açıkça Gösterilmesi (Voting Process)
  # Her bir kenar pikseli için tüm açılarda rho hesaplanır ve oylama matrisi artırılır.
  print('Oylama işlemi gerçekleştiriliyor...')
  for i in range(len(edge_x)):
    x = edge_x[i]
    y = edge_y[i]

    for t_idx in range(num_thetas):
      # Polar denklem: rho = x * cos(theta) + y * sin(theta)
      # Bellekte tutulan sin/cos değerleri kullanılarak hesaplama hızı artırılır
      rho = x * cos_thetas[t_idx] + y * sin_thetas[t_idx]

      # Rho değerini matris indeksine çevirmek için ofset eklenir (-diag_len kaydırılır)
      r_idx = int(rho + diag_len)

      # Akümülatördeki ilgili hücreye oy verilir (1 artırılır)
      accumulator[r_idx, t_idx] += 1

  # 5. Oylama Matrisini Resim Olarak Kaydetme ve İnceleme
  # Matris değerleri çok yüksek olabileceği için 0-255 aralığına normalize ediyoruz
  accumulator_normalized = cv2.normalize(
      accumulator, None, 0, 255, cv2.NORM_MINMAX, dtype=cv2.CV_8U
  )

  # Matrisi disk üzerine kaydetme
  cv2.imwrite('hough_accumulator.jpg', accumulator_normalized)
  print("Oylama matrisi 'hough_accumulator.jpg' olarak başarıyla kaydedildi.")

  # 6. Eşik Değeri Uygulayarak Doğruları Çizme
  threshold = (
      120  # Bu eşik değerini resminizin yoğunluğuna göre ayarlayabilirsiniz
  )
  img_custom_hough = img.copy()

  # Eşik değerini aşan akümülatör hücrelerini (tepe noktalarını) bulma
  r_idxs, t_idxs = np.where(accumulator > threshold)

  for r_idx, t_idx in zip(r_idxs, t_idxs):
    rho = rhos[r_idx]
    theta = thetas[t_idx]

    a = np.cos(theta)
    b = np.sin(theta)
    x0 = a * rho
    y0 = b * rho

    # Doğruyu görselleştirmek için büyük uzantılar hesaplanır
    x1 = int(x0 + 1000 * (-b))
    y1 = int(y0 + 1000 * (a))
    x2 = int(x0 - 1000 * (-b))
    y2 = int(y0 - 1000 * (a))

    # Bulunan doğruları mavi renkte çiziyoruz (BGR: 255, 0, 0)
    cv2.line(img_custom_hough, (x1, y1), (x2, y2), (255, 0, 0), 1)

  # Sonuçları Colab üzerinde gösterme
  print('Oylama Matrisi (Accumulator Görüntüsü):')
  cv2_imshow(accumulator_normalized)

  print('Özel Hough Dönüşümü Sonucu (Mavi Doğrular):')
  cv2_imshow(img_custom_hough)
```
